# ProXtal-LM — Interactive Metrics Explorer

An interactive plotting tool for exploring training metrics with full
widget-based controls: series toggles, title/axis editing, colour & style
pickers, legend/grid toggles, axis-limit sliders, and export to image.

**Requirements:**
```
pip install matplotlib ipywidgets pandas numpy
# or: uv pip install matplotlib ipywidgets pandas numpy
```

**Backend:** Use `%matplotlib widget` for best interactivity (pan/zoom
built into the canvas). Falls back to `%matplotlib inline` if
`ipympl` is not installed.

````python
# Set the matplotlib backend — %matplotlib widget is ideal but needs ipympl.
# If ipympl is not installed, fall back to inline.
try:
    import ipympl  # noqa: F401
    get_ipython().run_line_magic('matplotlib', 'widget')
except ImportError:
    get_ipython().run_line_magic('matplotlib', 'inline')
    print("Tip: install ipympl (`pip install ipympl`) for pan/zoom support.")
````

## Multi-Run Comparison

Compare metrics across all checkpoint runs. Columns are prefixed with the
run name so every series is individually togglable.

````python
import glob

csv_files = sorted(glob.glob('../checkpoints/*/training_metrics.csv'))
run_paths = {path.split('/')[-2]: path for path in csv_files}

print(f"Found {len(run_paths)} runs: {list(run_paths.keys())}")

if len(run_paths) > 1:
    multi_plotter = launch_multi(run_paths)
else:
    print("Only one run found — use the single-run explorer above.")
````
